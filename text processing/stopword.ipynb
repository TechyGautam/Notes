# Removing StopWords

````python
para = """
Natural Language Processing (NLP) is a fascinating field of study that focuses on the interaction between computers and human language. It involves the development of algorithms and models that enable machines to understand, interpret, and generate human language in a way that is both meaningful and useful. NLP has a wide range of applications, including sentiment analysis, machine translation, chatbots, and information retrieval. By leveraging techniques from linguistics, computer science, and artificial intelligence, NLP aims to bridge the gap between human communication and computer understanding, ultimately enhancing our ability to interact with technology in a more natural and intuitive manner."""
````

````python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

stop_words = set(stopwords.words('english'))


def remove_stopwords(text):
    words = word_tokenize(text)

    filtered_words = [
        word
        for word in words
        if word.lower() not in stop_words
    ]

    return ' '.join(filtered_words)


def lemmatization(text):
    lemmatizer = WordNetLemmatizer()

    words = word_tokenize(text)

    lemmatized_words = [
        lemmatizer.lemmatize(word, pos='v')
        for word in words
    ]

    return ' '.join(lemmatized_words)


new_para = remove_stopwords(para)

result = lemmatization(new_para)

print(result)
````

````python
print(result)
````

# 1. What are Stopwords?

**Stopwords** are words that occur very frequently in a language and are often considered less useful for certain NLP tasks.

Examples in English:

```text
the
is
a
an
and
or
of
to
in
on
for
with
was
are
```

For example:

```text
"The cat is sitting on the table."
```

Possible stopwords:

```text
the
is
on
the
```

After removing them:

```text
cat sitting table
```

---

# 2. Simple Definition

> **Stopword removal is the process of removing words that are considered unnecessary for a particular NLP task.**

The important phrase is:

> **"for a particular NLP task"**

Stopwords are **not universally useless words**.

This is very important.

---

# 3. Why Do We Remove Stopwords?

Consider a large document:

```text
The cat is on the table and the cat is near the window.
```

Words such as:

```text
the
is
on
and
```

may occur many times.

But for some tasks, the more informative words are:

```text
cat
table
near
window
```

Removing unnecessary words can:

- Reduce vocabulary size
- Reduce data size
- Reduce computation
- Make traditional NLP processing faster
- Focus models on potentially informative words

---

# 4. Real-Life Example

Imagine you want to search a library for books about:

```text
Machine Learning
```

You don't necessarily care about words like:

```text
the
is
a
of
```

You care more about:

```text
machine
learning
```

So stopword removal can simplify the text.

---

# 5. Example

Original sentence:


Possible stopwords:

```text

is

to
```

After removal:

```text

```

The sentence is grammatically incomplete, but for some NLP tasks the important content remains.

---

# 6. Important: Stopword Removal is Task-Dependent

This is one of the **most important concepts**.

You should NOT blindly remove every stopword.

Consider:

```text
I do not like this movie.
```

If you remove:

```text
I
do
not
```

you might get:

```text
like movie
```

Now the meaning has changed.

Original:

```text
I do not like this movie.
```

Meaning:

```text
Negative
```

After careless stopword removal:

```text
I like this movie.
```

Meaning:

```text
Positive
```

So:

> **Removing stopwords can sometimes destroy important meaning.**

---

# 7. Stopwords Are Not Always Useless

A word can be a stopword for one task and useful for another.

Example:

```text
not
```

In some traditional stopword lists, it may be included.

But in:

```text
I do not like this product.
```

`not` is extremely important.

Therefore:

```text
Stopword ≠ useless word
```

Instead:

```text
Stopword = word that may be removed for a specific task
```

---

# 8. Stopword List

A **stopword list** is a collection of words that we consider candidates for removal.

Example:

```python
[
    "the",
    "is",
    "a",
    "an",
    "and",
    "or",
    "of",
    "to",
    "in",
    "on"
]
```

During preprocessing, we check each token against this list.

---

# 9. How Stopword Removal Works

Suppose:

```text
"The cat is sleeping on the sofa."
```

### Step 1: Tokenization

```text
[
    "The",
    "cat",
    "is",
    "sleeping",
    "on",
    "the",
    "sofa"
]
```

### Step 2: Check Stopword List

```text
The       → stopword
cat       → keep
is        → stopword
sleeping  → keep
on        → stopword
the       → stopword
sofa      → keep
```

### Step 3: Remove Stopwords

```text
cat
sleeping
sofa
```

# 12. Basic Python Example

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

sentence = "This is a simple NLP example"

words = sentence.lower().split()

filtered_words = []

for word in words:
    if word not in stop_words:
        filtered_words.append(word)

print(filtered_words)
```

Possible output:

```text
['simple', 'nlp', 'example']
```

# 13. Shorter Python Version

The same thing can be written using list comprehension:

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

sentence = "This is a simple NLP example"

words = sentence.lower().split()

filtered_words = [
    word for word in words
    if word not in stop_words
]

print(filtered_words)
```

---

# 14. Why Use `set()`?

Instead of:

```python
stop_words = stopwords.words("english")
```

we often use:

```python
stop_words = set(stopwords.words("english"))
```

Why?

Checking membership in a set is generally faster:

```python
word in stop_words
```

For a large number of tokens, this can make a difference.

---

# 15. Stopwords With Tokenization

In real NLP preprocessing, we usually tokenize properly instead of using only:

```python
sentence.split()
```

Example:

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

sentence = "This is a simple NLP example."

stop_words = set(stopwords.words("english"))

words = word_tokenize(sentence)

filtered_words = [
    word for word in words
    if word.lower() not in stop_words
]

print(filtered_words)
```

Possible output:

```text
['simple', 'NLP', 'example', '.']
```

Notice:

```text
.
```

is not necessarily a stopword.

Punctuation removal is a **separate preprocessing step**.

### Stopwords

```text

```

### Punctuation

```text

```

Therefore:

```text
Stopword Removal ≠ Punctuation Removal
```

They may both happen during preprocessing, but they solve different problems.

# 17. Stopwords vs Tokenization

Tokenization:

> Break text into smaller units.

Example:

```text

```

↓

```text
["I", "love", "NLP"]
```

Stopword removal:

```text
["I", "love", "NLP"]
```

↓

```text
["love", "NLP"]
```

So:

```text
Tokenization
     ↓
Stopword Removal
```

A common pipeline is:

```text
Text
 ↓
Tokenization
 ↓
Stopword Removal
```

# 19. Custom Stopword List

You don't have to use only the default list.

Suppose your dataset contains:

```text
company
company
company
company
```

and `company` provides no useful information for your particular task.

You could create:

```python
custom_stopwords = {
    "the",
    "is",
    "a",
    "and",
    "company"
}
```

Then:

```python
filtered_words = [
    word for word in words
    if word.lower() not in custom_stopwords
]
```

---

# 20. Why Custom Stopwords Can Be Better

Different datasets have different characteristics.

For example, suppose you are analyzing customer reviews:

```text

```

Maybe:

```text

```

are useful.

But your dataset might contain a repeated company name:

```text

```

thousands of times.

If `XYZ` does not help your particular analysis, you might add it to your custom stopword list.

# 23. Advantages

## 1. Reduces Vocabulary

Suppose:

```text
Original vocabulary = 50,000 words
```

After removing unnecessary frequent words:

```text
Vocabulary < 50,000
```

Potentially reducing the feature space.

---

## 2. Reduces Computation

Fewer tokens/features can mean less processing.

---

## 3. Reduces Noise

Some extremely frequent words may provide little useful information for certain tasks.

---

## 4. Useful for Traditional NLP

Especially with representations such as:

```text
Bag of Words
TF-IDF
```

---

# 24. Disadvantages

## 1. Can Remove Important Information

Example:

```text
not
never
no
```

These can completely change meaning.

---

## 2. Can Hurt Sentiment Analysis

Example:

```text

```

Removing:

```text

```

can change the interpretation.

---

## 3. Can Hurt Question Answering

Words that look unimportant may be necessary to understand the exact meaning of a question.

---

## 4. Can Hurt Language Understanding

Natural language depends heavily on grammar and context.

Removing function words blindly can damage that information.

# 27. Stopwords with TF-IDF

TF-IDF already reduces the importance of words that occur frequently across documents.

For example:

```text
the
is
and
```

often receive low TF-IDF values when they appear in many documents.

Therefore, with TF-IDF:

> Stopword removal is not always necessary.

You should evaluate the effect on your dataset.

---

# 28. Important Modern NLP Note

In modern NLP, especially with:

```text
BERT
GPT-style models
Transformers
```

you should **not automatically remove stopwords**.

Why?

Because modern models use context.

Consider:

```text
I like this movie.
```

versus:

```text
I do not like this movie.
```

The word:

```text
not
```

is extremely important.

If preprocessing removes it:

```text
I like this movie.
```

you destroy important contextual information.

Therefore:

> Modern transformer pipelines generally keep the original words/tokens unless there is a specific reason to remove them.

# 34. Final Summary

**Stopword removal** is a text preprocessing technique used to remove selected words that may not contribute useful information to a particular NLP task.

Key points:

- Stopwords are usually very frequent words.
- Common examples include `the`, `is`, `and`, `of`.
- They can reduce vocabulary and computational work.
- Stopwords are **task-dependent**.
- They are not universally useless.
- Negation words such as `not`, `never`, and `no` can be extremely important.
- Custom stopword lists can be created.
- Stopwords are different from punctuation.
- They are particularly relevant in traditional NLP pipelines using BoW/TF-IDF.
- Modern transformer-based NLP generally should not blindly remove them.

### The Golden Rule ⭐

> **Never remove stopwords just because they are common. Remove them only when you know they are not useful for your specific NLP task.**
