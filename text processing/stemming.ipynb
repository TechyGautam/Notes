# What is Stemming?

**Stemming** is the process of reducing a word to its **root form (stem)** by removing prefixes or suffixes.

The resulting word **may or may not be a valid English word**.

### Simple Definition

> Stemming is a rule-based technique that removes prefixes or suffixes from words to obtain a common base form.

---

# Why Do We Need Stemming?

Imagine these words:

```
play
playing
played
plays
player
```

For humans, all of these are related to **play**.

But for a computer:

```
play ≠ playing ≠ played ≠ plays
```

Computer treats them as different words.

This increases:

- Vocabulary size
- Memory usage
- Computational cost

Stemming converts them into one common stem.

```
play
playing
played
plays

↓

play
```

Now the computer understands that these words are closely related.

# Why is Stemming Important in NLP?

Many NLP algorithms only care about the **meaning**, not the grammatical form.

Instead of treating

```
study
studies
studying

```

as four different words,

we use

```

```

(Porter Stemmer output)

This improves:

- Search Engines
- Text Classification
- Spam Detection
- Sentiment Analysis
- Recommendation Systems

# How Stemming Works

Stemming follows predefined **rules**.

It does **NOT** understand grammar.

Example rules:

```
Remove "ing"

playing

↓

play
```

Another rule

```
Remove "ed"

worked

↓

work
```

Another

```
Remove "ly"

quickly

↓

quick
```

Rules are applied one after another.

---

# Working Process

```
Sentence

↓

Tokenization

↓

Individual Words

↓

Stemming

↓

Root Words

↓

Machine Learning Model
```

Example

```
"I am studying NLP"

↓

["I","am","studying","NLP"]

↓

["I","am","studi","NLP"]
```

---

# Important Point

Stemming **does not care** whether the resulting word is a real dictionary word.

Example

```
studies

↓

studi
```

There is no English word called **studi**.

Still,

Porter Stemmer returns

```
studi
```

because it follows rules only.

---

# Rule-Based Algorithm

A stemming algorithm contains many rules.

Example:

```
ing → remove

playing

↓

play
```

```
ed → remove

worked

↓

work
```

```
ly → remove

quickly

↓

quick
```

```
ies → replace by i

studies

↓

studi
```

Thousands of such rules exist.

# Popular Stemming Algorithms

## 1. Porter Stemmer ⭐⭐⭐⭐⭐

Most popular.

Developed by:

```
Martin Porter (1980)
```

Characteristics

- Fast
- Rule-based
- Most commonly used
- Produces non-dictionary words

Example

```
studies

↓

studi
```

---

## 2. Snowball Stemmer ⭐⭐⭐⭐☆

Improved version of Porter Stemmer.

Advantages

- More accurate
- Supports multiple languages
- Faster

Languages supported:

- English
- French
- German
- Spanish
- Italian
- Dutch
- Russian
- etc.

---

## 3. Lancaster Stemmer ⭐⭐⭐⭐☆

Most aggressive stemmer.

Example

```
maximum

↓

maxim
```

Sometimes it over-stems.

Example

```
running

↓

run

```

Good.

But

```
general

↓

gen
```

May lose meaning.

---

# Comparison

| Stemmer | Speed | Accuracy | Aggressiveness |
|----------|--------|-----------|----------------|
| Porter | High | Good | Medium |
| Snowball | High | Better | Medium |
| Lancaster | Very High | Lower | High |

# Stemming Sentence

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()

sentence = "The boys are playing football"

words = sentence.split()

stemmed = []

for word in words:
    stemmed.append(ps.stem(word))

print(stemmed)
```

Output

```
['the', 'boy', 'are', 'play', 'footbal']
```

Notice

```
football

↓

footbal
```

This shows stemming only follows rules.

It doesn't understand language.

# Stemming vs Lemmatization

| Feature | Stemming | Lemmatization |
|----------|-----------|---------------|
| Uses Grammar | ❌ No | ✅ Yes |
| Uses Dictionary | ❌ No | ✅ Yes |
| Speed | Faster | Slower |
| Accuracy | Lower | Higher |
| Output | May not be real word | Real dictionary word |

Example

```
Studies

Stemming

↓

studi

Lemmatization

↓

study
```

# Interview Questions

### Q1. What is stemming?

Reducing words into their stem by removing prefixes or suffixes using predefined rules.

---

### Q2. Is stemming grammar-aware?

No.

---

### Q3. Can stemming produce invalid English words?

Yes.

Example

```
study

↓

studi
```

---

### Q4. Which stemming algorithm is most popular?

Porter Stemmer.

---

### Q5. Difference between stemming and lemmatization?

Stemming uses rules only.

Lemmatization uses grammar + dictionary.

# Common Mistakes

❌ Thinking stem is always a dictionary word.

Wrong.

```
studies

↓

studi
```

---

❌ Assuming stemming understands meaning.

It doesn't.

Only rules.

---

❌ Using stemming where high accuracy is needed.

For applications like machine translation or question answering, **lemmatization** is usually preferred.

---

# Best Practices

✅ Use Porter Stemmer for learning.

✅ Use Snowball Stemmer in many production cases.

✅ Use Lemmatization when preserving word meaning is important.

---

# Summary

- Stemming reduces words to a common stem.
- It is a rule-based process.
- The output may not be a valid English word.
- It helps reduce vocabulary size and speeds up NLP pipelines.
- Popular algorithms include Porter, Snowball, and Lancaster.
- Stemming is fast but less accurate than Lemmatization.
