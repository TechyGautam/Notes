**Corpus** means paragraphs. It is a collection of text data that can be used for various natural language processing tasks, such as tokenization, sentiment analysis, and text classification. In this case, the corpus consists of a single string containing multiple sentences.



```python
corpus = "Hello, my name is John!!.I am a software engineer.I love programming in Python's."

print(corpus)
```

#### We Use `from nltk.tokenize import sent_tokenize`
---> it is used to convert paragraph into list of sentences  or a sentences into a list of word
- `sent_tokenize` for gettting sentences.
- `word_tokenize` for getting words.

```python
from nltk.tokenize import sent_tokenize, word_tokenize 

corpus = """Artificial Intelligence is transforming the world. Machine Learning is a subset of AI.
Natural Language Processing enables computers to understand human language.
Deep Learning powers modern applications like ChatGPT, image recognition, and speech recognition.
"""
```

```python
## sentence tokenization
sent_list = sent_tokenize(corpus)
print("Sentence Tokenization:")
print(sent_list)
print("------------------------------")
for i in sent_list:
    print(i)
```

```python
word_list = word_tokenize(corpus)
print("Word Tokenization:")
print(word_list)
```

```python
print(word_tokenize("tanishkgautam19@gmail.com"))
print("tanishkgautam19@gmail.com".split())
```
