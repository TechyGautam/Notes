```python
corpus = """Artificial Intelligence is changing the way people work and learn.
Many students are learning machine learning because they want to build intelligent applications.
However, learning new technologies can sometimes be challenging and time-consuming."""
```

```python
from nltk.tokenize import sent_tokenize ,word_tokenize
from nltk.stem import PorterStemmer , WordNetLemmatizer
from nltk.corpus import stopwords

# Tokenization
sent_token = sent_tokenize(corpus)
word_token = word_tokenize(corpus)

print(sent_token)
print(word_token)

# Stemming and lemmatization
stemming = PorterStemmer()
lemmatizer = WordNetLemmatizer()
stem_lst = []
root_lst = []
for i in word_token:
    stem_lst.append(stemming.stem(i))
    root_lst.append(lemmatizer.lemmatize(i ,pos="v"))

print(stem_lst)
print(root_lst)


# removing stop words

stop_words = set(stopwords.words("english"))
final_lst = []
for i in root_lst:
    if i not in stop_words:
        final_lst.append(i)

print(" ".join(final_lst))
```
