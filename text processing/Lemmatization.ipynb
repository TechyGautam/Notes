# 1. What is Lemmatization?

**Lemmatization** is the process of converting a word into its **base or dictionary form**, called a **lemma**.

Unlike stemming, lemmatization tries to understand the **actual linguistic meaning and grammatical form** of a word.

### Simple Definition

> Lemmatization converts different forms of a word into its meaningful dictionary base form.

Examples:

```text
running  → run
ran      → run
better   → good
studies  → study
children → child
```

The important thing is that the result is generally a **valid word**.

---

# 2. What is a Lemma?

A **lemma** is the canonical/dictionary form of a word.

Examples:

```text
running
runs
ran
run

        ↓

       run
```

Here:

```text
run = lemma
```

Another example:

```text
children
child's
child

        ↓

      child
```

So:

```text
Original word → Lemma
```

---

# 3. Why Do We Need Lemmatization?

Consider:

```text
The boy is running.
The boys ran yesterday.
The boy runs every morning.
```

A computer may initially see:

```text
boy
boys
running
ran
runs
```

But semantically:

```text
boy
boys

↓

boy
```

and

```text
running
ran
runs
run

↓

run
```

Lemmatization reduces these different grammatical forms to their common base forms.

This can help NLP systems:

- Reduce vocabulary
- Improve text normalization
- Improve search
- Improve text matching
- Reduce unnecessary word variations
- Preserve linguistic meaning better than stemming


If you search:

# 5. How Lemmatization Works

Lemmatization is more sophisticated than simple suffix removal.

A simplified pipeline is:

```text
Input Word
    ↓
Tokenization
    ↓
Identify grammatical information
    ↓
Look up linguistic information
    ↓
Determine base form
    ↓
Lemma
```

For example:

```text
"running"

     ↓

Part of Speech = Verb

     ↓

Linguistic analysis

     ↓

"run"
```

# 6. Stemming vs Lemmatization

This is one of the **most important NLP concepts**.

Consider:

```text
studies
```

### Stemming

```text
studies → studi
```

Why?

Because the stemmer may simply remove/modify suffixes.

### Lemmatization

```text
studies → study
```

Why?

Because lemmatization knows that:

```text
studies
```

is an inflected form of:

```text
study
```

---

# 7. Main Difference

### Stemming

Think:

> "How can I cut this word down?"

### Lemmatization

Think:

> "What actual dictionary word does this word represent?"

This is the easiest way to remember the difference.

# 9. Why POS Matters

**POS = Part of Speech**

It tells us whether a word is:

- Noun
- Verb
- Adjective
- Adverb
- etc.

This is important because the same word can have different meanings/forms depending on its grammatical role.

For example:

```text
better
```

can be related to:

```text
good
```

when used as an adjective.

But a system needs linguistic information to make the correct interpretation.

Therefore, **POS information can improve lemmatization**.

---

# 10. Example of POS

Consider:

```text
I am running.
```

Here:

```text
running = Verb
```

Lemma:

```text
run
```

Now:

```text
I bought running shoes.
```

Here:

```text
running = Adjective
```

Its grammatical role is different.

Therefore, **context and POS can matter when determining a lemma**.

# 12. Basic Example

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = [
    "cats",
    "dogs",
    "running",
    "studies"
]

for word in words:
    print(word, "->", lemmatizer.lemmatize(word))
```

Possible output:

```text
cats -> cat
dogs -> dog
running -> running
studies -> study
```

Notice something interesting:

```text
running -> running
```

Why didn't it become:

```text
run
```

?

Because the lemmatizer may treat `running` as a noun by default.

This leads to one of the most important concepts:

# POS Tagging

---

# 13. Lemmatization with POS

If we tell the lemmatizer that:

```text
running = verb
```

then it can produce:

```text
running → run
```

Example:

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(
    lemmatizer.lemmatize("running", pos="v")
)
```

Output:

```text
run
```

Here:

```text
v = verb
```

---

# 14. Common POS Codes in WordNet

| Code | Meaning |
|---|---|
| `n` | Noun |
| `v` | Verb |
| `a` | Adjective |
| `r` | Adverb |
| `s` | Adjective satellite |

Examples:

```python
lemmatizer.lemmatize("running", pos="v")
```

```text
run
```

And:

```python
lemmatizer.lemmatize("better", pos="a")
```

```text
good
```

---

# 15. Why `pos` Is Important

Compare:

```python
lemmatizer.lemmatize("running")
```

with:

```python
lemmatizer.lemmatize("running", pos="v")
```

The first one may return:

```text
running
```

while the second gives:

```text
run
```

Because the second explicitly tells the system:

```text
running = verb
```

Therefore:

> Good lemmatization often depends on correct POS information.

- Word forms

# 26. When Should You Use Stemming?

Use stemming when:

- Speed is important
- Exact linguistic accuracy isn't critical
- You need simple normalization
- Search/matching is the main objective
- You are working with very large text collections

Example:

```text
Search Engine
```

If the user searches:

```text
connect
```

you may want related forms such as:

```text
connected
connecting
connection
```

to be considered.

---

# 27. When Should You Use Lemmatization?

Use lemmatization when:

- Meaning matters
- Linguistic correctness matters
- You need readable normalized words
- You are performing linguistic analysis
- You want better handling of irregular word forms

---

# 28. Important Modern NLP Note

Don't assume that every modern NLP project **must** perform lemmatization.

Traditional NLP pipelines often use:

```text
Tokenization
→ Stopword Removal
→ Stemming/Lemmatization
→ TF-IDF / BoW
→ ML Model
```

But modern transformer-based models such as BERT-style models generally use their own tokenization and contextual representations.

So:

> Lemmatization is an important NLP concept, but it is not automatically required in every modern NLP pipeline.

Wrong.

POS information can significantly affect the result.

Not necessarily.

If speed is more important:


If linguistic correctness matters:

```text
Lemmatization
```

may be preferable.

### ❌ Mistake 4: Thinking lemmatization is mandatory before ML

No.

- Dataset

Stemming usually removes word endings using rules, while lemmatization uses linguistic information and lexical resources to obtain a meaningful base form.

```text
Noun
Verb
Adjective
Adverb
```
